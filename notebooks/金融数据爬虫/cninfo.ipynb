```python
import pandas as pd
import requests



url = 'https://webapi.cninfo.com.cn/api/stock/p_stock2303'
payload = {
    'scode': '000001,688521',
    'rdate': '20250930',
    'type': '071001',
    '@column': 'SECCODE,SECNAME,F038N,F046N,F055N',
}


'''
enckey使用aes-cbc 加密
填充模式是 pkcs7
key 和 iv 一样 1234567887654321
加密的数据是当前的时间戳，9位的
'''
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad
import base64
import time

date_time = str(int(time.time())).encode()
key = b'1234567887654321'
iv  = b'1234567887654321'
cipher = AES.new(key, AES.MODE_CBC, iv)
padded_data = pad(date_time, AES.block_size, style='pkcs7')
encrypted_data = cipher.encrypt(padded_data)
enckey = base64.b64encode(encrypted_data).decode('utf-8')



headers = {
'accept':'application/json, text/javascript, */*; q=0.01',
'accept-enckey': enckey,
'accept-encoding':'gzip, deflate, br, zstd',
'accept-language':'zh-CN,zh;q=0.9,en;q=0.8',
'connection':'keep-alive',
'content-length':'0',
'host':'webapi.cninfo.com.cn',
'origin':'https://webapi.cninfo.com.cn',
'referer':'https://webapi.cninfo.com.cn/',
'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/54.0.2840.99 Safari/537.36',
'x-requested-with':'XMLHttpRequest',
          }


r = requests.post(url, data=payload, headers = headers)
r.text
```
