```python
import requests_html
import pandas as pd
from tqdm import tqdm
from requests_html import HTMLSession

session = HTMLSession()
dict = {
    '跑车':['9',26],
    '街车':['8',109],
    '踏板':['3',57],
    '复古':['13',16],
    '巡航':['4',26],
    '拉力':['11',14],
    '跨骑':['1',1],
    '旅行':['10',8],
    'MINI':['15',3],
    '越野':['5',22],
    '弯梁':['2',6],
    '三轮':['12',7],
    '其他':['14',4],
}

total_df = pd.DataFrame()
for category in dict:
    for page in tqdm(range(1, dict[category][1] + 1)):
        url = f'https://www.58moto.com/conditions/{page}-1-{dict[category][0]}-0-0-0-0-0-0-0-0-9-9-0-0-21-0-1050-1-0-0-0-0.html'
        r = session.get(url)
        brand_list = [x.text.split(' ')[0] for x in r.html.xpath('.//div[@class="brand"]')]
        model_list = [' '.join(x.text.split(' ')[1:]) for x in r.html.xpath('.//div[@class="brand"]')]
        name_list = [x.text for x in r.html.xpath('.//div[@class="brand"]')]
        price_list = [x.text for x in r.html.xpath('.//div[@class="price ellipsis"]')]
        zipped = list(zip(brand_list, model_list, name_list, price_list))
        df = pd.DataFrame(zipped, columns=['brand','model','name','price'])
        df['类型'] = category
        total_df = pd.concat([total_df,df])
```
