```python
import requests

def get_concept_reasons(page):
    url = 'https://datacenter.eastmoney.com/securities/api/data/v1/get'
    payload = {
        'reportName':'RPT_F10_CORETHEME_BOARDTYPE',
        'columns':'SECUCODE,SECURITY_NAME_ABBR,BOARD_NAME,SELECTED_BOARD_REASON',
        'quoteColumns':'f3~05~NEW_BOARD_CODE~BOARD_YIELD',
        'filter':'(IS_PRECISE="1")',
        'pageNumber': page,
        'pageSize':'',
        'sortTypes':'1',
        'sortColumns':'BOARD_RANK',
        'source':'HSF10',
        'client':'PC',
        'v':'05379500850666044',
    }
    r = requests.get(url, params = payload)
    df = pd.DataFrame(r.json()['result']['data']).drop(columns=['BOARD_YIELD'])
    return df

total_df = pd.DataFrame()
for page in range(1,5):
    df = get_concept_reasons(page)
    total_df = pd.concat([total_df,df])
total_df.columns = ['证券代码','股票名称','概念名称','概念解析']
```

```python
total_df.columns = ['证券代码','股票名称','概念名称','概念解析']
total_df['概念名称'] = total_df['概念名称'].str.replace('概念','')
total_df = total_df.drop_duplicates()
```
