```python
import requests
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)


energy_type = {
    1: ['燃油','燃油车'],
    4: ['纯电','新能源'],
    5: ['插混','新能源'],
    6: ['增程式','新能源']
}

rank_df = pd.DataFrame()
for i in energy_type:
    # url = 'https://car-web-api.autohome.com.cn/car/search/searchcar?searchtype=0&pageindex=1&pagesize=3000&orderid=40'
    url = f'https://cars.app.autohome.com.cn/carext/recrank/all/getrecranklistpageresult2?from=2&pm=2&pluginversion=11.58.5&model=1&channel=0&pageindex=1&pagesize=5000&typeid=1&subranktypeid=1&levelid=0&energytype={i}&price=0-9000&date=2024-02_2024-04'
    r = requests.get(url)

    df = pd.DataFrame(r.json()['result']['list'])
    df['sales'] = df['rightinfo'].apply(pd.Series)['righttextone']
    df['energy_type'] = energy_type[i][0]
    df['energy'] = energy_type[i][1]
    df = df[['seriesid','seriesname','priceinfo','sales','energy_type','energy']]
    rank_df = pd.concat([rank_df,df])
    
rank_df['sales'] = pd.to_numeric(rank_df['sales'], errors="coerce")
```

```python
rank_df = pd.DataFrame(rank_df.groupby(['seriesid','seriesname','priceinfo','sales','energy'])['energy_type'].apply(list)).sort_values(['sales'],ascending = False).reset_index()
```
